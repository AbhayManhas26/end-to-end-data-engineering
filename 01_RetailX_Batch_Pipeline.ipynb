```python
import pandas as pd
import numpy as np
import os
from datetime import datetime, timedelta

print("Libraries imported successfully!")
```

```python
# GENERATED RANDOM CUSTOMER DATA
customers = pd.DataFrame({
    "customer_id": range(1001, 1101),
    "customer_name": [f"Customer_{i}" for i in range(1, 101)],
    "city": np.random.choice(
        ["Delhi", "Mumbai", "Bangalore", "Pune", "Chennai"],
        100
    ),
    "registration_date": pd.date_range(
        start="2026-01-01",
        periods=100,
        freq="D"
    )
})

customers.head()
```

```python
# SAVING DATA
os.makedirs("../data/raw", exist_ok=True)

customers.to_csv(
    "../data/raw/customers.csv",
    index=False
)

print("Customer dataset saved successfully!")
# It automatically navigates:
# - .. → one folder back from notebooks
# - data/raw → into your project's raw data folder.
```

```python
# GENERATED RANDOM PRODUCT DATA
products = pd.DataFrame({
    "product_id": range(501, 551),
    "product_name": [f"Product_{i}" for i in range(1, 51)],
    "category": np.random.choice(
        ["Electronics", "Clothing", "Home", "Accessories"],
        50
    ),
    "price": np.random.randint(200, 50000, 50)
})

products.head()
```

```python
products.to_csv("../data/raw/products.csv", index=False)

print("Product dataset saved successfully!")
```

```python
# GENERATED RANDOM ORDER
orders = pd.DataFrame({
    "order_id": range(10001, 11001),
    "customer_id": np.random.choice(customers["customer_id"], 1000),
    "product_id": np.random.choice(products["product_id"], 1000),
    "quantity": np.random.randint(1, 5, 1000),
    "order_date": [
        datetime(2026, 11, 1) + timedelta(days=int(x))
        for x in np.random.randint(0, 26, 1000)
    ]
})

orders.head()
```

```python
orders.to_csv("../data/raw/orders.csv", index=False)

print("Order dataset saved successfully!")
```

```python
# GENERATED PAYMENT DATA
payments = pd.DataFrame({
    "payment_id": range(20001, 21001),
    "order_id": orders["order_id"],
    "payment_status": np.random.choice(
        ["Success", "Failed", "Pending"],
        1000,
        p=[0.85, 0.10, 0.05]
    ),
    "payment_method": np.random.choice(
        ["UPI", "Credit Card", "Debit Card", "Net Banking"],
        1000
    )
})

payments.head()
```

```python
payments.to_csv("../data/raw/payments.csv", index=False)

print("Payment dataset saved successfully!")
```

```python
# GENERATED RANDOM SHIPPING DATA
shipping = pd.DataFrame({
    "shipping_id": range(30001, 31001),
    "order_id": orders["order_id"],
    "shipping_status": np.random.choice(
        ["Delivered", "In Transit", "Delayed"],
        1000,
        p=[0.75, 0.20, 0.05]
    ),
    "delivery_days": np.random.randint(1, 8, 1000)
})

shipping.head()
```

```python
shipping.to_csv("../data/raw/shipping.csv", index=False)

print("Shipping dataset saved successfully!")
```

```python
# Validate our five source datasets
# Load all five datasets
import os
import pandas as pd

data_path = "../data/raw"

customers = pd.read_csv(f"{data_path}/customers.csv")
products = pd.read_csv(f"{data_path}/products.csv")
orders = pd.read_csv(f"{data_path}/orders.csv")
payments = pd.read_csv(f"{data_path}/payments.csv")
shipping = pd.read_csv(f"{data_path}/shipping.csv")

print("All five datasets loaded successfully!")
```

```python
# Check record counts
datasets = {
    "Customers": customers,
    "Products": products,
    "Orders": orders,
    "Payments": payments,
    "Shipping": shipping
}

for name, df in datasets.items():
    print(f"{name}: {len(df)} records")
```

```python
# Check missing values
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
# check duplicates records
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
import boto3

bucket_name = "retailx-data-engineering-2026"

s3 = boto3.client(
    "s3",
    region_name="ap-south-1"
)

s3.head_bucket(Bucket=bucket_name)

print("Successfully connected to RetailX S3!")
```

```python
# Upload our five datasets
files = [
    "customers.csv",
    "products.csv",
    "orders.csv",
    "payments.csv",
    "shipping.csv"
]

for file in files:
    local_path = f"../data/raw/{file}"
    s3_key = f"raw/{file}"

    s3.upload_file(
        local_path,
        bucket_name,
        s3_key
    )

    print(f"Uploaded: {file}")
```
