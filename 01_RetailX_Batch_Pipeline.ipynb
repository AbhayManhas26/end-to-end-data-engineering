```python
import boto3

bucket_name = "retailx-data-engineering-2026"

s3 = boto3.client(
    "s3",
    region_name="ap-south-1"
)

s3.head_bucket(Bucket=bucket_name)

print("Successfully connected to RetailX S3!")
```

```python
# Retrieve data directly from AWS S3
datasets_s3 = {}

files = [
    "customers.csv",
    "products.csv",
    "orders.csv",
    "payments.csv",
    "shipping.csv"
]

for file in files:
    response = s3.get_object(
        Bucket=bucket_name,
        Key=f"raw/{file}"
    )

    datasets_s3[file.replace(".csv", "")] = pd.read_csv(
        response["Body"]
    )

    print(f"Retrieved: {file}")
```

```python
# Verify the retrieved data
for name, df in datasets_s3.items():
    print(f"{name}:{len(df)}records")
```

```python
# Validating the processing data in S3

for name, df in datasets_s3.items():
    print(f"\n--- {name.upper()} ---")
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
```

```python
# Inspecting the dataset
for name, df in datasets_s3.items():
    print(f"\n========== {name.upper()} ==========")
    display(df.head(5))
    print("\nData types:")
    print(df.dtypes)
```

```python
# coverting the datatype of :-
# customers.registration_date
# orders.order_date
datasets_s3["customers"]["registration_date"] = pd.to_datetime(
    datasets_s3["customers"]["registration_date"]
)

datasets_s3["orders"]["order_date"] = pd.to_datetime(
    datasets_s3["orders"]["order_date"]
)
# Verify 
print(datasets_s3["customers"].dtypes)
print(datasets_s3["orders"].dtypes)
```

```python
# Validating table relation
customers = datasets_s3["customers"]
products = datasets_s3["products"]
orders = datasets_s3["orders"]
payments = datasets_s3["payments"]
shipping = datasets_s3["shipping"]

print("Invalid customer IDs:",
      (~orders["customer_id"].isin(customers["customer_id"])).sum())

print("Invalid product IDs:",
      (~orders["product_id"].isin(products["product_id"])).sum())

print("Invalid payment order IDs:",
      (~payments["order_id"].isin(orders["order_id"])).sum())

print("Invalid shipping order IDs:",
      (~shipping["order_id"].isin(orders["order_id"])).sum())
# Check whether IDs exist; 
# ~ reverses True/False to find invalid IDs, and sum() counts them (True=1, False=0).
```

```python
# DATA TRANSFORMATION
# Join Orders with Customers and Products

# Join orders with customer details
orders_enriched = orders.merge(
    customers,
    on="customer_id",
    how="left",
    validate="many_to_one"
)

# Join the result with product details
orders_enriched = orders_enriched.merge(
    products,
    on="product_id",
    how="left",
    validate="many_to_one"
)

display(orders_enriched.head())
print("Total records:", len(orders_enriched))
print("Total columns:", len(orders_enriched.columns))
# Understand the important parameters:
# - merge() — combines two DataFrames.
# - on — specifies the common column used to join.
# - how="left" — keeps every record from the left DataFrame (orders), even if no matching record exists on the right.
# - validate="many_to_one" — checks that each customer/product ID appears only once in its respective master table.
```

```python
# Add payment and shipping details

# Join payment details
orders_enriched = orders_enriched.merge(
    payments,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# Join shipping details
orders_enriched = orders_enriched.merge(
    shipping,
    on="order_id",
    how="left",
    validate="one_to_one"
)

display(orders_enriched.head())

print("Total records:", len(orders_enriched))
print("Total columns:", len(orders_enriched.columns))
# validate="one_to_one" ensures each order has at most one matching payment record and one matching shipping record, with unique order IDs on both sides.
```

```python
# Calculate total order amount
orders_enriched["total_amount"] = (
    orders_enriched["quantity"] * orders_enriched["price"]
)

display(
    orders_enriched[
        ["order_id", "quantity", "price", "total_amount"]
    ].head()
)
```

```python
# Validate the transformed dataset
print("Total records:", len(orders_enriched))
print("Total columns:", len(orders_enriched.columns))

print("Missing values:", orders_enriched.isnull().sum().sum())
print("Duplicate rows:", orders_enriched.duplicated().sum())

print("\nTotal sales value:", orders_enriched["total_amount"].sum())
```

```python
# Save the processed dataset locally
# save the transformed data separately from the raw data.
import os

# Create processed folder if it doesn't exist
os.makedirs("../data/processed", exist_ok=True)

# Save transformed dataset
orders_enriched.to_csv(
    "../data/processed/retailx_orders_processed.csv",
    index=False
)

print("Processed dataset saved successfully!")
# Why index=False?
# We don't want Pandas to save its row index as an unnecessary extra column in our CSV.
```

```python
# Step 1 — Create a database connection
import duckdb

# Create/connect to RetailX Data Warehouse
con = duckdb.connect(
    "../retailx_warehouse.duckdb"
)

print("RetailX Data Warehouse connected!")
```

```python
# Step 2 — Load the processed CSV into DuckDB
# Load processed CSV into DuckDB table
con.execute("""
    CREATE OR REPLACE TABLE retailx_orders AS
    SELECT *
    FROM read_csv_auto(
        '../data/processed/retailx_orders_processed.csv'
    )
""")

print("Data loaded into RetailX Warehouse!")
```

```python
# Step 3 — Verify the warehouse table
result = con.execute("""
    SELECT COUNT(*)
    FROM retailx_orders
""").fetchone()

print("Records in warehouse:", result[0])
```

```python
# check the total sales value by product category
sales_by_category = con.execute("""
    SELECT category,
        SUM(total_amount) AS total_sales
        from retailx_orders
        GROUP BY category
        ORDER BY total_sales DESC
    """).df()
display(sales_by_category)
```

```python
# Monthly sales analysis
monthly_sales = con.execute("""
    SELECT
        DATE_TRUNC('month', order_date) AS month,
        SUM(total_amount) AS total_sales
    FROM retailx_orders
    GROUP BY month
    ORDER BY month
""").df()

display(monthly_sales)
```

three useful business metrics from our warehouse:
1. Total orders
2. Total revenue
3. Average order value (AOV)

```python
business_metrics = con.execute("""
    SELECT
        COUNT(DISTINCT order_id) AS total_orders,
        SUM(total_amount) AS total_revenue,
        AVG(total_amount) AS average_order_value
    FROM retailx_orders
""").df()

display(business_metrics)
```

```python
import pandas as pd
import duckdb
import matplotlib.pyplot as plt

con = duckdb.connect("../retailx_warehouse.duckdb")

print("Connected to RetailX Data Warehouse!")
```

```python
# Verifying the table
con.execute("""
    SELECT COUNT(*) 
    FROM retailx_orders
""").fetchone()
```

```python
# first dashboard chart: Revenue by category
category_sales = con.execute("""
    SELECT
        category,
        SUM(total_amount) AS total_revenue
    FROM retailx_orders
    GROUP BY category
    ORDER BY total_revenue DESC
""").df()

category_sales
plt.figure(figsize=(9, 5))

plt.bar(
    category_sales["category"],
    category_sales["total_revenue"]
)

plt.title("RetailX - Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Monthly revenue trend
monthly_sales = con.execute("""
    SELECT
        DATE_TRUNC('month', order_date) AS month,
        SUM(total_amount) AS total_revenue
    FROM retailx_orders
    GROUP BY month
    ORDER BY month
""").df()

monthly_sales
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales["month"],
    monthly_sales["total_revenue"],
    marker="o"
)

plt.title("RetailX - Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Payment status breakdown
payment_summary = con.execute("""
    SELECT
        payment_status,
        COUNT(*) AS total_orders
    FROM retailx_orders
    GROUP BY payment_status
    ORDER BY total_orders DESC
""").df()

payment_summary
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    payment_summary["total_orders"],
    labels=payment_summary["payment_status"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("RetailX - Payment Status Distribution")
plt.show()
```

```python
# Shipping status breakdown
shipping_summary = con.execute("""
    SELECT
        shipping_status,
        COUNT(*) AS total_orders
    FROM retailx_orders
    GROUP BY shipping_status
    ORDER BY total_orders DESC
""").df()

shipping_summary
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    shipping_summary["shipping_status"],
    shipping_summary["total_orders"]
)

plt.title("RetailX - Shipping Status Distribution")
plt.xlabel("Shipping Status")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```

```python
# Retrieve KPI metrics
kpis = con.execute("""
    SELECT
        COUNT(*) AS total_orders,
        SUM(total_amount) AS total_revenue,
        AVG(total_amount) AS average_order_value
    FROM retailx_orders
""").df()

# Create dashboard layout
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Revenue by category
axes[0, 0].bar(
    category_sales["category"],
    category_sales["total_revenue"]
)
axes[0, 0].set_title("Revenue by Category")
axes[0, 0].tick_params(axis="x", rotation=20)

# 2. Monthly revenue
axes[0, 1].plot(
    monthly_sales["month"],
    monthly_sales["total_revenue"],
    marker="o"
)
axes[0, 1].set_title("Monthly Revenue Trend")
axes[0, 1].tick_params(axis="x", rotation=30)

# 3. Payment status
axes[1, 0].pie(
    payment_summary["total_orders"],
    labels=payment_summary["payment_status"],
    autopct="%1.1f%%"
)
axes[1, 0].set_title("Payment Status")

# 4. Shipping status
axes[1, 1].bar(
    shipping_summary["shipping_status"],
    shipping_summary["total_orders"]
)
axes[1, 1].set_title("Shipping Status")
axes[1, 1].tick_params(axis="x", rotation=20)

# Dashboard heading
fig.suptitle("RETAILX - BUSINESS INTELLIGENCE DASHBOARD",
             fontsize=16, fontweight="bold")

# KPI summary
fig.text(
    0.5, 0.92,
    f"Total Orders: {kpis.iloc[0]['total_orders']:.0f}  |  "
    f"Revenue: ₹{kpis.iloc[0]['total_revenue']:,.0f}  |  "
    f"Average Order Value: ₹{kpis.iloc[0]['average_order_value']:,.2f}",
    ha="center",
    fontsize=12
)

plt.tight_layout(rect=[0, 0, 1, 0.88])
plt.show()
```
